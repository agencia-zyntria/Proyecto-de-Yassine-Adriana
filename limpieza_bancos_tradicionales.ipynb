```python
# Identificar filas donde todos los precios están vacíos
columnas_precios = ["Close", "High", "Low", "Open"]

filas_sin_precios = datos_bancos_limpios.loc[
    :, datos_bancos_limpios.columns.get_level_values(0).isin(columnas_precios)
].isnull().all(axis=1)

# Eliminar esas filas
datos_bancos_limpios = datos_bancos_limpios.loc[~filas_sin_precios]
```

```python
for banco in ["BBVA.MC", "CABK.MC", "UNI.MC"]:
    error = (
        (datos_bancos_limpios[("High", banco)] <
         datos_bancos_limpios[("Low", banco)])
    )
    print(banco, "Errores:", error.sum())
```

```python
precios_cierre = datos_bancos_limpios["Close"]
```

```python
precios_comunes = precios_cierre.dropna()
```

```python
print("Primera fecha:", precios_comunes.index.min())
print("Última fecha:", precios_comunes.index.max())
print("Dimensiones:", precios_comunes.shape)
```

```python
precios_mensuales = precios_comunes.resample("ME").last()
precios_mensuales.head()
```

```python
precios_mensuales = precios_mensuales.dropna()

precios_mensuales.head()
```
